## Problem context

Target function : Exclusive or function
Returns one only when strictly one of the input set is one.
0 + 1 => 1
1 + 1 => 0
0 + 0 => 0

Targeted for exact representation of the dataset, making 100 percent pure target hit for the ideal function.

<center> Loss Function  </center>

$$
    Loss = \frac{1}{n} \sum{(f^*(x) - f(x;{\theta}))^2}
$$

<center> Linear Model</center>

$$
    f(x;w,b) = x^Tw + b
$$

*Breif into least square* -> 
We project the target vector into feature space,
Projection onto the feature space is the best possible linear relation
with the target vector through the feature space.

With Least square implementation we get weights as zero with bias 0.5 making all of the result to go with 0.5 as the prediction


```python
# Linear regression setup for  XOR problem
import numpy as np
'''
Bias Term implementation detail
A small but reasonable stuff with building linear model, 
Without bias dummy columns with one, least square wont return the bias terms,
The only reason the model wont be optimum it would be best stretch of weigths form origin
'''
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

bias = np.ones(X.shape[0])
X_stack = np.column_stack(
    [X, 
    bias]
    )

y = np.array([0, 1, 1, 0])

# Making simple Least square based regression model
np.linalg.lstsq(X_stack,y) # With weights ~ 0 and bias 0.5 | best model prediction
```

```python
# Hidden-layer | Neurals building block

''' 
With real neural internals

With random weights initialized
1. Forward pass [ computing prediction with weights ]
2. Computing loss with initial predictions
3. Autograd + Backpropagation of the main function
4. Optimization step for the weights

Going the down hill with each step, tuning weights based on loss function,
directions defined through its differentiation steps.

implementation with manual forward passing with 
'''
W = np.array([
    [1, 1],
    [1, 1]
])


c = np.array([0, -1])

# Output-layer parameters
w = np.array([1, -2])
b = 0

# Forward propagation
Z = X @ W + c
H = np.maximum(0, Z)
predictions = H @ w + b

print("Pre-activations:\n", Z)
print("Hidden outputs:\n", H)
print("Predictions:", predictions)
print("Targets:    ", y)
```
