# Editing DataFrame

Simple dictionary based editing into the core data frame unit

```python
import numpy as np
import pandas as pd

df = pd.DataFrame(
    np.random.rand(10,4), index=range(0,10), columns=list("ABCD")
)

df
```

```python
# New columns and DF editing
df['Addition'] = 100 * df['A']
'''
If the name does not exist then it would be added to the dataframe
we can also edit the existing data frame with simple dict like flow

Since editing is really straight forward
we must need to have a copy of the original dataset before doing something into

We need to read weather its original or its copy
'''
df.loc[2, 'Addition'] = 50
# lock with location based on a condition and edit them
case = df['A'] < 0.5
df.loc[case, 'A']=0 # Need to be precise about what are we doing into

df # New edited data frame with new column
```

```python
new_entries = [
    "vrindavan",
    "mathura",
    "agra",
    "firozabad",
    "akhbarpur",
    "indexing",
    "selection",
    "droping",
    "editing",
    "pandas"
]
df['places'] = new_entries

df
```

```python
# Conditional selection with isin function
filter_topics = df['places'].isin(['pandas','editing','droping','indexing'])
'''
Indexing with reference

Need to know which to target
where to lock in :)
'''
print(filter_topics) # Specifically made with places while declaring
df.loc[filter_topics]
```

```python
# Deleting columns
'''
Removing rows or columns by specific label
with inplace operation it would do into the DB itself

by default returns a copy of the df
we can also suppress error through error parameter
it would be raised when the column or label is not found
'''
df.drop('A', axis=1)
```
