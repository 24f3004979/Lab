```python
# loc and iloc based precise extraction from data frame

import pandas as pd
import numpy as np

df = pd.DataFrame(np.random.rand(10,4), index=list('abcdefghij'), columns=list('ABCD'))
df
```

```python
# Making precise selection with loc wayout with columns

df.loc['d', 'A':'C'] # Precise selection of the dataset
```

```python
# using lambda function with selection of data through a data selection :)
selector = lambda df: df["A"]  > 0.8
df.loc[selector]
```
