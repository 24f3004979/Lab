# Foundations with torch

torch made by facebook, it serves to work as a framework to do machine learning tasks, it helps us to lift the all heavy lifting with numbers and there transformations. With awesome gpu support it can fast up calculations with parallel processing power.

With torch we can learn how to use that tool, to design the mathematical models on top of the state of the art library tools which it provides us starting with tensor an generic mathematical object defined for doing mathematical operations into for doing calculation at good scale.

```python
# Tensor basics with torch
import torch
'''
We can reffer the official documentation, but when we get stuck or need a help doing something with numbers
Reading docs is the most awesome way to use the tool, trust me, its better then asking gpt, its dopamine high
'''

# Declare tensor with python standard data structure
tensor = torch.tensor([1,2,3])

tensor # simlpe torch.tensor object declared | Powerfull to do a lot of stuff under the hood | refer the docs for detailed journey
```

```python
# Foundational Data basics with tensor behaviour

# Referencing variable behaviour
new_tensor = tensor + 10

print(f"Reffered variable {new_tensor}")

# editing original source
tensor[1] = 49

print(tensor)
print(new_tensor) # This should be not modified is it ?
''' Upon close inspection the reffered
variable with adding 10 have also changed thus we need to be cautious with defining variable flow '''

# Referencing variables and copy pathways
'''With just clone gradients can flow back in reference web, detatch is required | Find out why ?'''
z_tensor = tensor.clone().detach()  # Clean Deep copy
print(z_tensor)

# editing source
tensor[1] = 103
print("The source tensor with edits ", tensor)
print("The reference with deep copy for other state", z_tensor)

# With same we can convert tensor to numpy and bought back them find out how to get pythonic elements through tensor ?
```

```python
# Foundational mathematical functions

x = torch.rand(size=(2,2)) # random matrix with 2 X 2 dimension

determinant = torch.det(x.float())
determinant = determinant.item()
reshaped_tensor = x.reshape(1,4) # Its a view of original
'''Explore the other foundations with Linear algebra operations we can do with tensors'''
# Making edits with the source
x[1,1] = 3 # changed at position 1,1

print(x)
print(reshaped_tensor)
```

```python
# Making reproducible code with torch
random_seed = 9
torch.manual_seed(random_seed)  # Minecraft world generates with seed :)

random_tensor = torch.rand(size=(10,10))

print(random_tensor) # with no matter how many times you run output would be same
```

```python
# switching to gpu/cpu
device = "cuda"
tensor = torch.tensor([1,2,3])
print(tensor, tensor.device) # Where its present ?

tensor = tensor.to(device)
print(tensor, tensor.device)
print(type(tensor))  # Torch tensor for the gpu acceleration

# Geting GPU Elements into numpy
element = tensor.cpu().numpy()
print(element, type(element))
```
